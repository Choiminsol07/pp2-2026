# 12장 상속

- 상속에 대한 이론과 실습

```python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return (
            f"Maker: {self.maker}, Model: {self.model}, "
            f"Color: {self.color}, Price: {self.price}"
        )


class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def getBatterySize(self):
        return self.batterySize

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery Size: {self.batterySize}"


class Truck(Car):
    def __init__(self, maker, model, color, price, payload):
        super().__init__(maker, model, color, price)
        self.payload = payload

    def __str__(self):
        return f"{super().__str__()}, Payload: {self.payload}"


car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())

ecar1 = ElectricCar("Tesla", "Model S", "Red", 80000, 100)
print(ecar1)
ecar1.setBatterySize(98)
print(ecar1.getDesc())

truck1 = Truck("Ford", "F-150", "Black", 30000, 1000)
print(truck1)
print(truck1.getMaker())

print(type(car1))
print(type(ecar1))
print(type(truck1))

print(isinstance(car1, Car))
print(isinstance(ecar1, Car))
print(isinstance(truck1, Car))

print(isinstance(truck1, Car))
```

# 다중 상속 : 2개 이상의 부모 클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1=10):
        self.attr1 = v1

    def display1(self):
        print(f"{self.attr1}")


class Base2:
    def __init__(self, v2=20):
        self.attr2 = v2

    def display2(self):
        print(f"{self.attr2}")


class C(Base1, Base2):
    def __init__(self, v1, v2, v3):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3

    def print(self):
        print(f"{self.attr1},{self.attr2},{self.attr3}")
        self.display1()
        self.display2()
        Base1.display1(self)
        Base2.display2(self)


def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()

    c2 = C(4, 5, 6)
    c2.print()


if __name__ == "__main__":
    test()
```

# Overriding : 부모 함수(매소드) 재정의

```python
import math


class Shape:
    def __init__(self, name="도형"):
        self.name = name

    def draw(self):
        print(f"{self.name}을(를) 그립니다.")

    def get_area(self):
        print(f"Shape: {self.name}의 면적을 계산합니다.")


class Circle(Shape):
    def __init__(self, radius=1, name="원"):
        super().__init__(name)
        self.radius = radius

    def draw(self):
        print(f"Circle: {self.name}을(를) 그립니다.")

    def get_area(self):
        print(f"Circle: {self.name}의 면적을 계산합니다.")
        return math.pi * self.radius ** 2


class Ractangle(Shape):
    def __init__(self, width=1, height=1, name="사각형"):
        super().__init__(name)
        self.width = width
        self.height = height

    def draw(self):
        print(f"Ractangle: {self.name}을(를) 그립니다.")

    def get_area(self) -> float:
        print(f"Rectangle: {self.name}의 면적을 계산합니다.")
        return self.width * self.height


def test():
    c = Circle(10)
    c.draw()
    print(c.get_area())

    r1 = Ractangle(width=5, height=10)
    r1.draw()
    print(r1.get_area())

    print("-------------------------------------")

    circles = {Circle(1), Circle(2), Circle(3)}

    for c in circles:
        c.draw()
        print(c.get_area())

    print("-------------------------------------")


if __name__ == "__main__":
    test()
```
