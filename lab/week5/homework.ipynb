# 12장 상속

- 상속이란? 기존의 클래스로부터 변수와 메소드를 상속받아 새로운 클래스를 파생하는 메커니즘
- 코드 재사용, 유지 및 보수 가능
- 부모 클래스(Parent Class): 다른 클래스에 기능을 물려주는 클래스
- 자식 클래스(Child Class): 부모 클래스의 기능을 물려받는 클래스

- type(): 객체가 어떤 자료형 또는 클래스로 만들어졌는지 확인하는 함수
- isinstance(): 객체가 특정 클래스의 객체인지 확인하는 함수. 상속 관계에 있는 부모 클래스까지 확인할 수 있다.

```python
# 학번 이름을 통한 상속 예시

class Student:
    def __init__(self):
        self.name = "최민솔"
        self.number = "202611845"

class MyStudent(Student):
    def show(self):
        print(self.number, self.name)

s = MyStudent()
s.show()
```

```python
# lab 은행의 이율 계산하기

class Bank():
    def getInterestRate(self):
        return 0.0

class BadBank(Bank):
    def getInterestRate(self):
        return 10.0

class NormalBank(Bank):
    def getInterestRate(self):
        return 5.0

class GoodBank(Bank):
    def getInterestRate(self):
        return 3.0

b1 = BadBank()
b2 = NormalBank()
b3 = GoodBank()

print("BadBank의 이자율: " + str(b1.getInterestRate()))
print("NormalBank의 이자율: " + str(b2.getInterestRate()))
print("GoodBank의 이자율: " + str(b3.getInterestRate()))
```

```python
# lab 직원과 메니저

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def getSalary(self):
        return self.salary

class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def getSalary(self):
        salary = super().getSalary()
        return salary + self.bonus

    def __str__(self):
        return "이름: " + self.name + "; 월급: " + str(self.salary) + "; 보너스: " + str(self.bonus)

kim = Manager("김철수", 20000000, 1000000)
print(kim)
```

- 메소드 오버라이딩(Method Overriding): 부모 클래스에서 정의한 메소드를 자식 클래스에서 다시 정의하여 사용하는 것

```python
class A:
    def hello(self):
        print("안녕하세요")

class B(A):
    def hello(self):
        print("반갑습니다")

a = A()
b = B()

a.hello()
b.hello()
```

- 클래스를 만들 때 별도로 부모 클래스를 지정하지 않아도 기본적으로 object 클래스를 상속받는다.

```python
# __repr__()은 객체를 개발자가 이해하기 쉬운 문자열로 표현하는 특별한 메소드다.

class Student:
    def __repr__(self):
        return "Student(최민솔, 202611845)"

s = Student()
print(repr(s))
```

```python
#__str__()은 Python에서 객체를 사람이 읽기 쉬운 문자열로 표현하는 특별한 메소드다.

class Student:
    def __str__(self):
        return "최민솔"

s = Student()
print(s)
```

- __str__과 __repr__ 모두 object 클래스의 메소드로써 읽이 쉬운 문자열로 표현할 때 사용하는 매소드다. 
- 출력할 때 __str__()은 최민솔 학생처럼 자연스럽게 표현하고, __repr__()은 Student('최민솔', '202611845')처럼 객체의 정보를 구체적으로 표현할 수 있다.

```python
# 1번
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"


class Point3D(Point):
    def __init__(self, x, y, z):
        super().__init__(x, y)
        self.z = z

    def __str__(self):
        return f"({self.x}, {self.y}, {self.z})"
```

```python
# 2번

class Address:
    def __init__(self, street, city):
        self.street = str(street)
        self.city = str(city)


class Person:
    def __init__(self, name, email):
        self.name = str(name)
        self.email = str(email)


class Contact(Address, Person):
    def __init__(self, name, email, street, city):
        Address.__init__(self, street, city)
        Person.__init__(self, name, email)
```

```python
# 3번

import math


class Function:
    def value(self, x):
        raise NotImplementedError


class Quadratic(Function):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def value(self, x):
        return self.a * x ** 2 + self.b * x + self.c

    def get_roots(self):
        d = self.b ** 2 - 4 * self.a * self.c

        if d < 0:
            return None
        elif d == 0:
            return (-self.b / (2 * self.a),)
        else:
            x1 = (-self.b + math.sqrt(d)) / (2 * self.a)
            x2 = (-self.b - math.sqrt(d)) / (2 * self.a)
            return x1, x2
```

```python
# 4번

import random


class Dice:
    def roll(self):
        return random.randint(1, 6)


class FraudDice(Dice):
    def roll(self):
        while True:
            n = super().roll()
            if n > 3:
                return n
```

```python
# 5번

class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Dog(Animal):
    pass


class Cat(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)
        self.breed = breed


def get_oldest_cat(*args):
    return max(args, key=lambda cat: cat.age)


cat1 = Cat("나비", 3, "페르시안")
cat2 = Cat("초코", 7, "코리안숏헤어")
cat3 = Cat("루루", 5, "샴")

oldest = get_oldest_cat(cat1, cat2, cat3)
print(f"가장 나이가 많은 고양이는 {oldest.age}살입니다.")
```

```python
# 6번 

class Department:
    def __init__(self, name):
        self.name = name
        self.courses = []

    def add_course(self, name, credit):
        course = Course(name, credit)
        self.courses.append(course)
        return course


class Course:
    def __init__(self, name, credit):
        self.name = name
        self.credit = credit
        self.students = []


class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.courses = []

    def enroll(self, course):
        self.courses.append(course)
        course.students.append(self)


dept = Department("컴퓨터")
math1 = dept.add_course("공업수학", 3)
math2 = dept.add_course("이산수학", 2)

kim = Student("Kim", 20200001)
kim.enroll(math1)
```

```python
# 7번

class Artist:
    def __init__(self, name):
        self.name = name
        self.albums = []


class Album:
    def __init__(self, title, artist, year):
        self.title = title
        self.artist = artist
        self.year = year
        self.tracks = []
        artist.albums.append(self)

    def add_track(self, song):
        self.tracks.append(song)


class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add_song(self, song):
        self.songs.append(song)


lee = Artist("Lee's Band")
album = Album("첫 번째 앨범", lee, 2020)
album.add_track("첫 번째 노래")
album.add_track("두 번째 노래")

playlist = Playlist("애창곡")

for song in album.tracks:
    playlist.add_song(song)
```
