# 10장 파일과 예외처리

- 파일: 보조기억장치에서 문서, 소리, 그림, 동영상과 같은 자료를 모아놓은 것.
- 논리적인 정보 단위
- open: 파일 열기
- close: 파일 닫기
- r: 읽기 모드
- w: 쓰기 모드
- a: 추가 모드
- r+: 읽기와 쓰기 모드 

- 예외처리: 프로그램을 실행하는 도중 오류가 발생했을 때 프로그램이 갑자기 종료되지 않도록 처리하는 방법

```python
# 파일에 내용 저장하기

with open("hello.txt", "w", encoding="utf-8") as file:
    file.write("안녕하세요")

# 파일 내용 읽기
with open("hello.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
```

```python
# lab 매출파일처리

infilename = input("입력 파일 이름: ")
outfilename = input("출력 파일 이름: ")
infile = open(infilename, "r") 
outfile = open(outfilename, "w")
sum = 0
count = 0
line = infile.readline()
while line != "" :
s = int(line) 
sum += s
count += 1
line = infile.readline()
outfile.write("총매출 = "+ str(sum)+"\n")
outfile.write("평균 일매출 = "+ str(sum/count ))
infile.close()
outfile.close()
```
