# 파일 매출 처리

- 일별 평균 매출과 총 매출을 계산한 후에 다른 파일에 출력하는 프로그램을 작성해보자.

```python
# 매출 파일 처리 (p.456)

# 입력 파일과 출력 파일 지정
infilename = "input.txt"
outfilename = "output.txt"

try:
    # 파일 열기
    infile = open(infilename, "r", encoding="utf-8")
    outfile = open(outfilename, "w", encoding="utf-8")

    # 합계와 개수
    total = 0
    count = 0

    # 파일에서 한 줄씩 읽기
    line = infile.readline()

    while line:
        line = line.strip()

        # 빈 줄은 건너뛰기
        if line:
            total += int(line)
            count += 1

        line = infile.readline()

    # 결과 저장
    outfile.write("합계: " + str(total) + "\n")
    outfile.write("개수: " + str(count) + "\n")

    # 파일 닫기
    infile.close()
    outfile.close()

    print("파일 처리가 완료되었습니다.")
    print("합계:", total)
    print("개수:", count)

except FileNotFoundError:
    print("오류: input.txt 파일을 찾을 수 없습니다.")
    print("input.txt 파일을 현재 작업 폴더에 넣어주세요.")

except ValueError:
    print("오류: input.txt에 숫자가 아닌 값이 들어 있습니다.")
```

### 정규식: 특정한 규칙을 가지고 있는 문자열들을 메타 문자를 이용하여 표현하는 수식, 문자열에서 특정한 규칙이나 패턴을 찾아내고, 검사하거나, 수정하는 데 사용한다.

```python
# 정규식을 이용하여 학번만 출력하기

import re

text = "학생 최민솔의 학번은 202611845입니다."

result = re.findall(r"\d+", text)

print(result)
```

```python
# lab 패스워드 검사 프로그램 (p.487)

import re
password = input("패스워드를 입력하세요: ")
flag = 0
while True:
    if (len(password)<8):
        flag = -1
        break
    elif not re.search("[a-z]", password):
        flag = -1
        break
    elif not re.search("[A-Z]", password):
        flag = -1
        break
    elif not re.search("[0-9]", password):
        flag = -1
        break
    elif not re.search("[_@$]", password):
        flag = -1
        break
    else:
        flag = 0
        print("유효한 패스워드")
        break

if flag == -1:
    print("유효한 패스워드가 아닙니다.")
```

### 예외처리: 프로그램 실행 중 오류가 발생했을 때 적절하게 대응하여 프로그램이 갑자기 종료되는 것을 방지하는 방법
- try:
-    예외가 발생할 수 있는 문장
- expect(오류):
-    예외를 처리하는 문장

```python
# 예외처리를 통해 학번 입력받기

try:
    student_id = input("학번을 입력하세요: ")

    if not student_id.isdigit():
        raise ValueError("학번은 숫자로 입력해야 합니다.")

    if len(student_id) != 9:
        raise ValueError("학번은 9자리여야 합니다.")

except ValueError as e:
    print("오류:", e)

else:
    print("학번:", student_id)

finally:
    print("프로그램을 종료합니다.")
```

```python
# 8번
with open('scores.txt', 'r', encoding='utf-8') as f:
    scores = [float(line.strip()) for line in f if line.strip()]

print("읽은 점수:", scores)
print("점수 개수:", len(scores))

if len(scores) > 0:
    average = sum(scores) / len(scores)

    with open('scores.txt', 'a', encoding='utf-8') as f:
        f.write(f'평균값: {average:.1f}\n')

    print(f'평균값: {average:.1f}')
else:
    print("점수 데이터가 없습니다.")
```
