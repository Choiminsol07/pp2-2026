# 파일 매출 처리

- 일별 평균 매출과 총 매출을 계산한 후에 다른 파일에 출력하는 프로그램을 작성해보자.

```python
# 매출 파일 처리 (p.456)

# 입력 파일 이름 , 출력 파일 이름 받기
infilename = input("입력 파일 이름:")
outfilename = input("출력 파일 이름:")

# 입력 , 출력 파일 열기
infile = open(infilename, "r")
outfile = open(outfilename, "w")

# 합계 , 횟수 변수 정의
sum = 0
count = 0

line = infule.readline()
while line != "":
    s = int(line)
    sum += s
    count += 1
    line = infile.readline()

outfile.write("총매출 = "+ str(sum)+"\n")
outfile.write("평균 일매출 = "+ str(sum/count))

infile.close()
outfile.close()
```

# 작업 폴더: python 파일 실행하는 위치

```python
with open("file.txt", "r", encoding="utf-8") as f:
    #for line in f:
     #  print(line.strip())

    line = f.readline().strip()
    while line != "":
        print(line)
        line = f.readline().strip()

    eline = f.readline().strip()
    print(eline)
    print(f.tell())

    f.seek(0,0)
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
    
```
