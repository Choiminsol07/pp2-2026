# 파일 매출 처리

- 일별 평균 매출과 총 매출을 계산한 후에 다른 파일에 출력하는 프로그램을 작성해보자.

```python
# 매출 파일 처리 (p.456)

# 입력 파일과 출력 파일 지정
infilename = "input.txt"
outfilename = "output.txt"

try:
    # 파일 열기
    infile = open(infilename, "r", encoding="utf-8")
    outfile = open(outfilename, "w", encoding="utf-8")

    # 합계와 개수
    total = 0
    count = 0

    # 파일에서 한 줄씩 읽기
    line = infile.readline()

    while line:
        line = line.strip()

        # 빈 줄은 건너뛰기
        if line:
            total += int(line)
            count += 1

        line = infile.readline()

    # 결과 저장
    outfile.write("합계: " + str(total) + "\n")
    outfile.write("개수: " + str(count) + "\n")

    # 파일 닫기
    infile.close()
    outfile.close()

    print("파일 처리가 완료되었습니다.")
    print("합계:", total)
    print("개수:", count)

except FileNotFoundError:
    print("오류: input.txt 파일을 찾을 수 없습니다.")
    print("input.txt 파일을 현재 작업 폴더에 넣어주세요.")

except ValueError:
    print("오류: input.txt에 숫자가 아닌 값이 들어 있습니다.")
```

# 작업 폴더: python 파일 실행하는 위치
