### 내장함수
- abs() 절댓값 반환 함수
- all() 전체 항목 T/F 함수
- any() 한개의 항목이라도 참이면 참 반환 함수
- bin() 정수의 이진 표현 반환 함수
- eval() 전달된 수식을 구문 분석하고 계산하는 함수
- sum() 합 함수
- len() 객체 길이 함수
- list() 리스트 생성 함수
- map() 반복 가능한 항목에 함수를 적용 후 결과 반환 함수
- dir() 객체가 가지고 있는 변수나 함수를 보여주는 함수
- complex(real,imag) 복소수 객체를 생성하는 함수
- max(), min() 가장 큰 정수, 가장 작은 정수 반환
- enumerate() 열거형 객체 반환
- filter() 특정 조건을 만족하는 요소만 뽑는 함수
- zip() 2개의 리스트를 하나로 묶어주는 함수

```python
# 내장함수를 이용한 학번 이름 출력하기

numbers = [202611845]
names = ["최", "민", "솔"]

id = str(max(numbers))
name = "".join(list(map(str, names)))

result = " ".join([id, name])

print(result)
print("학번 길이:", len(id))
print("이름 글자 수:", len(name))
print("학번의 이진수:", bin(int(id)))
print("모든 정보가 존재하는가?", all([id, name]))
print("학번 또는 이름이 존재하는가?", any([id, name]))
print("학생 정보:", list(zip(["학번", "이름"], [id, name])))
print("정보 열거:", list(enumerate([id, name])))
print("학번 최댓값:", max(numbers))
print("학번 최솟값:", min(numbers))
print("학번 절댓값:", abs(-int(id)))
print("복소수 예제:", complex(1, 0))
print("이름의 문자 목록:", list(name))
print("사용 가능한 문자열 메서드:", "upper" in dir(name))
print("계산 결과:", eval("1 + 2"))
print("짝수 숫자:", list(filter(lambda x: x % 2 == 0, [1, 2, 3, 4])))
```

### 람다식
- 람다식: 무명함수, 이름은 없고 몸체만 있는 함수이다. 주로 딱 한 번 사용되는 함수를 만드는 데 사용된다.
- 형태: lambda 매개변수들: 수식
- 람다식은 내장함수와 함께 사용되기도 한다.

```python
# lab 람다식으로 온도 변환하기 (p.519)

f_temp = [0,10,20,30,40,50]

c_temp = map(lambda x: (5.0/9.0) * (x - 32.0), f_temp)
print(list(c_temp))
```

```python
# lab 람다식으로 데이터 처리하기 (p.520)

orders = [["1","재킷",5,12000],
            ["2","바지",3,10000],
            ["3","셔츠",10,5000],
            ["4","모자",2,7000]]
result = list(map(lambda x: [x[0], x[1], x[2], x[3], x[2]*x[3]], orders))
print(result)
```

### 이터레이터와 제너레이터
- 이터레이터: for 루프와 함께 사용할 수 있는 여러 종류의 반복 가능한 객체, 데이터를 한 번에 전부 꺼내는 대신 필요할 때마다 하나씩 꺼낼 수 있다.
- 제너레이터: 키워드 yield를 사용하여서 함수로부터 반복 가능한 객체를 생성하는 하나의 방법, 값을 필요할 때 하나씩 만들어서 반환하는 특별한 이터레이터

```python
# 이터레이터를 통해 숫자를 하나씩 출력하기

numbers = [1, 2, 3]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
```

```python
# 제너레이터를 통해 숫자를 하나씩 만들기

def numbers():
    yield 1
    yield 2
    yield 3

for number in numbers():
    print(number)
```

### 모듈
- 모듈: 함수나 변수 또는 클래스들을 모아 놓은 파일
- copy, random, sys(인터프리터정보제공),time, calendar 등 다양한 형태의 모듈이 있다.

```python
# lab 동전 던지기 게임

import random
myList = ["head","tail"]

while (True):
    respone = input("동전 던지기 게임을 시작하시겠습니까? (y/n): ")
    if respone == "y":
        coin = random.choice(myList)
        print(coin)
    else:
        break
```
