## Tuple, Dictionary, Set 실습


byte, bytes, bytearray 공부

all(), any() 공부

```python
#byte란? 컴퓨터가 사용하는 가장 기본 단위

x = b'A'
print(x)
#A에 저장돼있는 숫자가 출력
```

```python
#bytes란? byte가 모인 것.

x = b'ABC'
#파이선 내에서 bytes는 시퀀스로 활용된다.
#byte 내에 3개의 byte가 순서대로 들어있으며 인덱스로 꺼낼 수 있다.

print(x[0])
#A에 저장돼있는 숫자가 출력

#str은 문자를 꺼내고 bytes는 숫자를 꺼내는 시퀀스!!
```

```python
#bytearray란? byte 데이터를 저장하며 내용을 수정할 수 있는 자료형

b = bytearray([65, 66, 67])
print(b)
#bytearray(b'ABC')로 출력, ABC가 각각 숫자로 저장돼있음

a = bytearray(b"ABC")
a[0] = 90
print(a)
#bytearray(b'ZBC')러 출력, A가 Z로 바뀜
```

```python

#all()란? 자료형에 잇는 모든 요소가 참이면 true를, 거짓이면 false를 반환하는 함수
#참 : True, 1, "hello", [1,2] 
#거짓: False, 0, "", []

all([1,2,3])
#True
all([1,0,3])
#False

#all()는 리스트, 튜플, 문자열, 집합, 딕셔너리 등에서도 사용 가능
```

```python
#any()란? 자료형에 있는 요소 중 하나만이라도 참이면 true를, 모두 거짓이면 false를 반환하는 함수
#참 : True, 1, "hello", [1,2] 
#거짓: False, 0, "", []

any([1,2,3])
#True
any([1,0,3])
#True
any([0,0,0])
#False

#any()는 리스트, 튜플, 문자열, 집합, 딕셔너리 등에서도 사용 가능
```
