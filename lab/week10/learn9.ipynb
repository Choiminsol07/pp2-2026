# GUI 프로그래밍

- GUI: graphical user interface (그래픽 사용자 인터페이스), 아이콘·버튼·창을 통해 컴퓨터와 상호작용할 수 있게 해준다.
- TUI: test user interface (텍스트 기바나 사용자 인터페이스), 터미널에서 문자와 키보드 입력으로 조작할 수 있다.
- 단순위쳇: Button(버튼), Canavs(그림판), Checkbutton(체크박스), Entry(입력칸), Label(글자표시), Message(여러 줄의 글상자)
- 윈도우(Window): 화면 안에서 프로그램이나 내용을 보여주는 네모난 영역(창)

```python
# 버튼 위젯

from tkinter import * # thiker 모듈 포함
window = Tk() # Tk()을 호출하면 Tk 클래스의 객체가 생성되면서 화면에 하나의 윈도우가 생성된다.

button = Button(window, text="클릭하세요!", bg="yellow", fg="blue", width=80, height=2) # 버튼 위젯을 생성한 후 전경색과 배경색, 크기를 설정한다.

button.pack() # 버튼 위젯을 윈도우에 배치한다.
window.mainloop() # 윈도우가 사용자 동작을 대기한다.
```

```python
# 압축 배치 관리자 : 위젯을 최대한 압축하여 상하로 배치한다.

from tkinter import *
window = Tk()

Button(window,text = "박스#1", bg = "red", fg = "white").pack()
Button(window,text = "박스#2", bg = "green", fg = "black").pack()
Button(window,text = "박스#3", bg = "blue", fg = "orange").pack()

window.mainloop()
```

```python
# 격자 배치 관리자 : 위쳇을 테이블 형태로 배치, 부모 윈도우는 테이블의 셀로 분할되고 각 위젯은 특정한 셀에 배치된다.

from tkinter import *
window = Tk()

b1 = Button(window,text = "박스#1", bg = "red", fg = "white")
b2 = Button(window,text = "박스#2", bg = "green", fg = "black")
b3 = Button(window,text = "박스#3", bg = "blue", fg = "orange")
b4 = Button(window,text = "박스#4", bg = "pink", fg = "blue")

b1.grid(row = 0, column = 0) # grid() : 위젯을 테이블의 특정한 셀에 배치한다. row : 행, column : 열
b2.grid(row = 0, column = 1)
b3.grid(row = 1, column = 0)
b4.grid(row = 1, column = 1)

window.mainloop()
```

```python
# 절대 위치 배치 관리자 : 위젯을 부모 윈도우의 특정한 좌표에 배치한다.

from tkinter import *

window = Tk()

b1 = Button(window, text="박스 #1", bg="red", fg="white")
b1.place(x=0, y=0) # place() : 위젯을 부모 윈도우의 특정한 좌표에 배치한다. x : x좌표 y : y좌표    
b2 = Button(window, text="박스 #2", bg="green", fg="black")
b2.place(x=20, y=30)
b3 = Button(window, text="박스 #3", bg="blue", fg="orange")
b3.place(x=40, y=60)

window.mainloop()
```

```python
# lab 숫자 추측 게임


from tkinter import *
import random

answer = random.randint(1, 100)

def guessing():
    try:
        guess = int(guessField.get())

        if guess < 1 or guess > 100:
            resultLabel["text"] = "1부터 100까지 입력하세요."
        elif guess > answer:
            resultLabel["text"] = "높음!"
        elif guess < answer:
            resultLabel["text"] = "낮음!"
        else:
            resultLabel["text"] = "정답입니다! 🎉"

        guessField.delete(0, END)

    except ValueError:
        resultLabel["text"] = "숫자를 입력하세요!"

def reset():
    global answer
    answer = random.randint(1, 100)
    resultLabel["text"] = "다시 한번 도전하세요!"
    guessField.delete(0, END)

window = Tk()
window.configure(bg="lightblue")
window.title("숫자 추측 게임")
window.geometry("350x200")

titleLabel = Label(
    window,
    text="숫자 게임에 오신 것을 환영합니다",
    bg="lightblue"
)
titleLabel.pack(pady=15)

guessField = Entry(window)
guessField.pack(pady=5)

tryButton = Button(
    window,
    text="시도",
    fg="blue",
    bg="lightgray",
    command=guessing
)
tryButton.pack(pady=3)

resetButton = Button(
    window,
    text="초기화",
    fg="blue",
    bg="lightgray",
    command=reset
)
resetButton.pack(pady=3)

resultLabel = Label(
    window,
    text="1부터 100 사이의 숫자를 입력하시오.",
    bg="lightblue"
)
resultLabel.pack(pady=5)

window.mainloop()
```

```python
# lab 가위바위보 게임

import random
from tkinter import *
window = Tk()
Label(window, text="선택하세요", font=("Helvetica", "16")).pack()
frame = Frame(window)

rock_image = PhotoImage(file="rock.png")
paper_image = PhotoImage(file="paper.png")
scissors_image = PhotoImage(file="scissors.png")

def pass_s():
    decide("가위")
def pass_r():
    decide("바위")
def pass_p():
    decide("보")

rock = Button(frame, image=rock_image, command=pass_r)
rock.pack(side="left")
paper = Button(frame, image=paper_image, command=pass_p)
paper.pack(side="left")
scissors = Button(frame, image=scissors_image, command=pass_s)
scissors.pack(side="left")

frame.pack()

Label(window,text="컴퓨터는 다음을 선택하였습니다.", font=("Helvetica", "16")).pack()
computer_image = Label(window, image=rock_image)
computer_image.pack()
output = Label(window, text="", font=("Helvetica", "16"))
output.pack()

def decide(human):
    computer = random.choice(["가위", "바위", "보"])
    if computer == "가위":
        computer_image["image"]=scissors_image
    elif computer == "바위":
        computer_image["image"]=rock_image
    else:
        computer_image["image"]=paper_image
    if (computer == "가위" and human == "보") or (computer == "바위" and human == "가위") or (computer == "보" and human == "바위"):
        output["text"]="컴퓨터 승리!"
    elif computer == human:
        output["text"]="무승부!"
    else:
        output["text"]="사용자 승리!"
    output["text"] += f" (사용자: {human}, 컴퓨터: {computer})"

window.mainloop()
```
