# GUI 프로그래밍

- GUI: graphical user interface (그래픽 사용자 인터페이스), 아이콘·버튼·창을 통해 컴퓨터와 상호작용할 수 있게 해준다.
- TUI: test user interface (텍스트 기바나 사용자 인터페이스), 터미널에서 문자와 키보드 입력으로 조작할 수 있다.
- 단순위쳇: Button(버튼), Canavs(그림판), Checkbutton(체크박스), Entry(입력칸), Label(글자표시), Message(여러 줄의 글상자)
- 윈도우(Window): 화면 안에서 프로그램이나 내용을 보여주는 네모난 영역(창)

```python
# 버튼 위쳇

from tkinter import * # thiker 모듈 포함
window = Tk() # Tk()을 호출하면 Tk 클래스의 객체가 생성되면서 화면에 하나의 윈도우가 생성된다.

button = Button(window, text="클릭하세요!", bg="yellow", fg="blue", width=80, height=2) # 버튼 위젯을 생성한 후 전경색과 배경색, 크기를 설정한다.

button.pack() # 버튼 위젯을 윈도우에 배치한다.
window.mainloop() # 윈도우가 사용자 동작을 대기한다.
```
